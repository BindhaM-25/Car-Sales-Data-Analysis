```python
df["Make"] = df["Make"].fillna(df["Make"].mode()[0])
df["Model"] = df["Model"].fillna(df["Model"].mode()[0])
df["Type"] = df["Type"].fillna(df["Type"].mode()[0])
df["Origin"] = df["Origin"].fillna(df["Origin"].mode()[0])
df["DriveTrain"] = df["DriveTrain"].fillna(df["DriveTrain"].mode()[0])
df
```

```python
df["EngineSize"] = df["EngineSize"].fillna(df["EngineSize"].median())
df["Cylinders"] = df["Cylinders"].fillna(df["Cylinders"].median())
df["Horsepower"] = df["Horsepower"].fillna(df["Horsepower"].median())
df["MPG_City"] = df["MPG_City"].fillna(df["MPG_City"].median())
df["MPG_Highway"] = df["MPG_Highway"].fillna(df["MPG_Highway"].median())
df["Weight"] = df["Weight"].fillna(df["Weight"].median())
df["Wheelbase"] = df["Wheelbase"].fillna(df["Wheelbase"].median())
df["Length"] = df["Length"].fillna(df["Length"].median())
```

```python
df=df.drop_duplicates()
df
```

```python
df.describe()
```

```python
df["Make"].unique()
df["Type"].unique()
df["Origin"].unique()
df
```

```python
df['MSRP'] = df['MSRP'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
df['Invoice'] = df['Invoice'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)


df['MSRP'] = pd.to_numeric(df['MSRP'], errors='coerce')
df['Invoice'] = pd.to_numeric(df['Invoice'], errors='coerce')


df['MSRP'] = df['MSRP'].fillna(df['MSRP'].mean())
df['Invoice'] = df['Invoice'].fillna(df['Invoice'].mean())
```
